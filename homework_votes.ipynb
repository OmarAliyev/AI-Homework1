```python

votes = ["apple", "banana", "apple", "orange",
         "banana", "apple", "apple", "orange"]

fruits = set(votes)

for fruit in fruits:
    print(fruit, "→", votes.count(fruit))

leader = max(fruits, key=votes.count)

print("Lider:", leader)

print("İlk 3 netice:")

counts = {}

for fruit in fruits:
    counts[fruit] = votes.count(fruit)

sorted_fruits = sorted(counts, key=counts.get, reverse=True)

for fruit in sorted_fruits[:3]:
    print(fruit, "→", counts[fruit])
```
