```python

math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

subjects = {
    "Riyaziyyat": math,
    "Fizika": physics,
    "İngilis dili": english
}

results = {}

for subject, scores in subjects.items():
    results[subject] = {
        "Avarage": sum(scores) / len(scores),
        "Minimum": min(scores),
        "Maxsimum": max(scores)
    }

for subject, values in results.items():
    print(
        subject,
        "| Avarage:", round(values["Avarage"], 2),
        "| Minimum:", values["Minimum"],
        "| Maxsimum:", values["Maxsimum"]
    )

best_subject = max(
    results,
    key=lambda subject: results[subject]["Avarage"]
)

print("Max score subject: ", best_subject)
```
