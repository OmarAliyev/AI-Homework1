```python

expenses = [1200, 1500, 800, 600, 1000, 400, 900]

total = sum(expenses)
average = total / len(expenses)
maximum = max(expenses)
minimum = min(expenses)

days_over_1000 = 0

for i in expenses:
    if i>1000:
        days_over_1000+=1


print("Total: ", total)
print("Avarege: ", round(average, 2))
print("Max: ", maximum)
print("Min: ", minimum)
print("Days over than 1000: ", days_over_1000)
```
